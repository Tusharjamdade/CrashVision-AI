```python
# Install Ultralytics
!pip install -q ultralytics

from ultralytics import YOLO
import os

# Path to dataset root
DATASET_PATH = "accident/data"   # Change if needed

# Verify structure
print("Train classes:", os.listdir(f"{DATASET_PATH}/train"))
print("Val classes:", os.listdir(f"{DATASET_PATH}/val"))
print("Test classes:", os.listdir(f"{DATASET_PATH}/test"))

# Load YOLOv8 Classification Model
model = YOLO("yolov8n-cls.pt")

# Train
results = model.train(
    data=DATASET_PATH,
    epochs=20,
    imgsz=224,
    batch=16,
    device="cpu",
    workers=0,
    project="accident_detection",
    name="yolov8_cls"
)

# Evaluate
metrics = model.val()

print("\nValidation Results:")
print(metrics)

# Save best model path
print("\nBest model:")
print("accident_detection/yolov8_cls/weights/best.pt")
```

```python
import cv2
from ultralytics import YOLO

model = YOLO(
    r"C:\Users\Owner\runs\classify\accident_detection\yolov8_cls\weights\best.pt"
)

cap = cv2.VideoCapture("video.mp4")

while True:
    ret, frame = cap.read()

    if not ret:
        break

    result = model.predict(frame, verbose=False)

    pred = result[0].names[result[0].probs.top1]
    conf = float(result[0].probs.top1conf)

    cv2.putText(
        frame,
        f"{pred} {conf:.2f}",
        (20,40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0,255,0),
        2
    )

    cv2.imshow("Accident Detection", frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
